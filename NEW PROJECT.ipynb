```python
# outlier analysis
Q1=df["Sales"].quantile(0.25)
Q3=df["Sales"].quantile(0.75)
IQR = Q3-Q1
lower_bound = Q1-1.5*IQR
upper_bound = Q3+1.5*IQR
outliers = df[(df["Sales"]<lower_bound)|(df["Sales"]>upper_bound)]
print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
print("Lower Bound:",lower_bound)
print("Upper Bound:",upper_bound)
print("Number of Sales Outliers:",len(outliers))
```

```python
outliers[["Order_ID","Product","Category","Quantity","Sales","Profit"]].head(10)
```

```python
outliers["Category"].value_counts()
```

The identified sales outliers appear to represent higher-value transactions rather than obvious data-entry errors,so they should be retained for analysis

```python
plt.figure(figsize=(8,6))
sns.histplot(df['Sales'],bins =30,kde=True)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.histplot(df["Profit"],bins= 30,kde=True)
plt.title("Profit distribution")
plt.xlabel("Profit")
plt.ylabel("Frequency")
plt.show()
```

```python
category_margin = (df.groupby("Category")[["Sales","Profit"]].sum())
category_margin["Profit_Margin"]=(category_margin["Profit"]/category_margin["Sales"]*100)
category_margin.sort_values("Profit_Margin",ascending=False)
```

```python
product_analysis=(df.groupby("Product")[["Sales","Profit"]].sum())
product_analysis["Profit_Margin"]=(product_analysis["Profit"]/product_analysis["Sales"]*100)
product_analysis.sort_values("Sales",ascending=False).head()
```

```python
product_analysis.sort_values("Profit_Margin",ascending=False).head(10)
```

```python
monthly_analysis = df.groupby(["Year","Month","Month_Name"])[["Sales","Profit"]].sum()
monthly_analysis
```

```python
monthly_analysis= monthly_analysis.reset_index()
```

```python
plt.figure(figsize=(10,5))
plt.plot(monthly_analysis["Month_Name"],monthly_analysis["Sales"],marker="o",label="Sales")
plt.plot(monthly_analysis["Month_Name"],monthly_analysis["Profit"],marker="o",label="Profit")
plt.title("Monthly Sales Vs Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.show()
```

FINAL INSIGHTS --------

North America had the highest sales among all regions.

Electronics was the highest-selling category.

Office Chair generated the highest sales among products.

Notebook Set had the highest profit margin.

December had the highest sales and profit, while February had the lowest.

Sales and Profit had a strong positive correlation (0.94).

Quantity and Profit also showed a positive correlation (0.72).

Discount and Profit had a weak negative correlation (-0.10).

Customer Satisfaction had almost no correlation with sales or profit.

The analysis found some high-value sales outliers, which were kept because they may represent genuine large transactions.
 The analysis shows that Sales and Profit were strongly related,with Electronics and Office Chair performing well in Sales.
  December was the strongest month,while customer satisfaction and discounts showed little relationship with overall profit.
